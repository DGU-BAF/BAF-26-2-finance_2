# 차입 규모·만기구조와 이자부담: 가설 확장 탐색 분석
실행일 2026-11-03. 사용자 승인: 가설을 바꾸는 방향과 차입비용 전달·영업실적 분해를 합쳐 추가 분석. 기존 EDA·회귀·원자료를 보존한다. 기존 비유의 결과를 본 이후의 탐색 분석이며 사전 확증 연구처럼 서술하지 않는다.

## 가설과 변경 이력 (회귀 실행 전 고정)
기존 질문: 사전 단기차입/자산이 높은 업종의 긴축기 ICR이 더 악화되는가?
확장 질문: 차입 총량과 단기 비중을 구분하면 이자부담·수익성·ICR의 변화가 다르게 나타나는가?
- H1: 총차입/자산과 유동비율을 고려할 때 사전 단기비중이 높을수록 부담금리 대용지표의 사후 증가가 크다(양의 방향).
- H2: 단기비중과 유동비율을 고려할 때 사전 총차입/자산이 높을수록 이자비용/전기자산의 사후 증가가 크다(양의 방향).
- RQ3: 두 차입 특성의 ICR·영업이익률 관계와 회계적 ICR 변화 구성을 함께 살펴본다. 결과를 본 뒤 유리한 방향으로 가설을 다시 쓰지 않는다.

기존 E=단기차입/자산 → Q=제공 차입 네 항목 합계/자산, S=단기 두 항목/차입 네 항목 합계. 둘은 사전 2018–20년 각 연도 비율의 평균. 연도별로 E=Q×S/100이지만 3년 평균에서는 평균의 곱이 아니므로 단순한 동일 모형 재표현은 아니다. L=기존 유동비율. 사전 기준·기간을 유지해 변수 정의를 분리한다.

모든 모형은 40업종, 2018–23, Post=2022–23, 2021 별도, 업종·연도FE. 모든 결과변수에 동일 설명변수 QP,SP,LP,QT,ST,LT. 전체 표본 사용, 극단값 제외·윈저화 없음. 실제 금리 모형은 이번에 추가하지 않는다. 원자재·환율·생산지수도 이번 분석에 추가하지 않는다.

## 변수 및 미확인 정의
D=단기차입+단기사채+장기차입+장기사채(십억원). 공식 전체 이자부채와 같다고 확인되지 않았고 유동성장기부채·리스부채 등의 포함 여부는 미확인이다. 서로 독립적인 항목이라는 잠정 해석에 기초한 구성 지표다. 이자비용과 차입의 정확한 집계 범위 역시 공식 산식 확인이 필요하다.
Y1 cost=100×이자비용/[(D_t+D_t-1)/2]. 평균잔액 선택은 연간 비용과 잔액을 대응시키기 위한 분석자 선택, 실제 약정금리가 아니다. Y2 burden=100×이자비용/전기자산. Y3 margin=100×영업이익/매출. Y4 icr=기존 공시 ICR%/100. 모두 이자비용·영업이익 등 금액 열이 동일 범위로 집계됐다는 잠정 전제가 있다.
제공 차입의존도 열은 Q로 사용하지 않는다. Q·S를 금액에서 직접 계산하고 공식 비율 열과의 대조 결과를 공개한다. 산업별 금액은 정수 십억원이므로 작은 분모에서 반올림 오차가 커질 수 있다. 담배 등 작은 차입 분모의 높은 cost도 삭제하지 않는다.

## 회계적 분해 (인과효과 아님)
업종별 2018–20년 영업이익·이자비용의 평균 A0,B0와 2022–23년 평균 A1,B1을 사용한다. R=A/B인 재구성 ICR의 변화에 대해 영업이익 기여=0.5×(A1−A0)×(1/B0+1/B1), 이자비용 기여=0.5×(A0+A1)×(1/B1−1/B0). 두 가능한 변경 순서의 평균이며 두 기여의 합이 R1−R0와 정확히 일치한다. 공식 연간 ICR의 평균 변화와 같지 않으며 이를 바꿔치기하지 않는다. 음의 영업이익에서는 이자비용 감소의 수치상 기여 방향이 직관과 달라질 수 있다. 인과적 매개효과 검정이 아니다.

## 문헌·공식 출처 확인
[Kitsul·Lang·Samadi(2023), FEDS Notes II절](https://www.federalreserve.gov/econres/notes/feds-notes/monetary-policy-tightening-and-debt-servicing-costs-of-nonfinancial-companies-20231201.html): “annualized interest expense scaled by total debt”. 관련 II·IV·V절 본문 확인. 비용/부채로 전달을 먼저 측정하는 접근을 참고했다. 본 연구는 해외 상장기업·분기자료이며 은행차입비중과 현금비율을 사용한다. 우리 단기비중·유동비율과 같지 않다. 원 연구의 기말부채 대비 비용과 우리 평균잔액 지표는 구분하며 같은 산식이라 주장하지 않는다. 원 연구 V절 EBITDA 기반 ICR도 우리의 영업이익 기반 ICR과 다르다.
원 XLSX 메타정보: TOPIC_DT_NICE_KISVALUE_03(안정성), _02(수익성), NICE평가정보/ValueSearch. 금액 단위 십억원을 원파일 헤더에서 확인. 메타정보에는 필요한 항목별 포함 범위·집계 산식이 없어 검증 완료로 표시하지 않는다. 사용자는 설명을 보유하지 않는다고 답했다. 제출용 확정 해석에는 제공기관의 산식 확인이 필요하다.

재현: Python3.12 + requirements.txt, inputs와 함께 이 노트북 폴더에서 전체 실행. 원자료에서 최초 패널 생성은 기존 01_analysis.ipynb; 여기서는 동결된 입력 금액으로 새 변수와 모형을 재계산한다.


```python
# 재무자료 연결
from pathlib import Path
import json,hashlib
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy import stats
from statsmodels.stats.multitest import multipletests
from IPython.display import display
from wildboottest.wildboottest import wildboottest
BASE = Path.cwd()
(BASE / 'tables').mkdir(exist_ok=True)
def save(x, n):
    x.to_csv(BASE / 'tables' / f'{n}.csv', index=False, encoding='utf-8-sig')
for m in json.loads((BASE/'inputs/manifest.json').read_text(encoding='utf-8')):
 assert hashlib.sha256((BASE/'inputs'/m['file']).read_bytes()).hexdigest()==m['sha256']
s = pd.read_csv(BASE / 'inputs/stability.csv')
p = pd.read_csv(BASE / 'inputs/profit.csv')
orig = pd.read_csv(BASE / 'inputs/panel.csv')
f=s.merge(p.drop(columns=['업종']),on=['업종코드','연도'],validate='one_to_one').rename(columns={'업종코드':'code','연도':'year','업종':'industry'}).sort_values(['code','year'])
f=f[f.code.isin(orig.code.unique())].copy()

# 차입 구성과 비용 지표 계산
f['D']=f[['단기차입','단기사채','장기차입','장기사채']].sum(axis=1,min_count=4)
f['short'] = f['단기차입'] + f['단기사채']
f['Qa'] = 100 * f.D / f['자산총계']
f['Sa'] = 100 * f.short / f.D
f['La'] = 100 * f['유동자산'] / f['유동부채']
f['lagD'] = f.groupby('code').D.shift()
f['lagA'] = f.groupby('code')['자산총계'].shift()
f['lagyear'] = f.groupby('code').year.shift()
f['cost'] = 100 * f['이자비용'] / ((f.D + f.lagD) / 2)
f['cost_end'] = 100 * f['이자비용'] / f.D
f['burden'] = 100 * f['이자비용'] / f.lagA
f['margin'] = 100 * f['영업이익'] / f['매출액']
f['icr'] = f['이자보상비율'] / 100
f['icr_amount'] = f['영업이익'] / f['이자비용']

# 분모와 관측기간 검산
pre = f[f.year.between(2018, 2020)]
assert pre.groupby('code').size().eq(3).all()
av=pre.groupby('code')[['Qa','Sa','La']].mean().rename(columns={'Qa':'Q','Sa':'S','La':'L'})
t=f[f.year.between(2018,2023)].merge(av,on='code',validate='many_to_one').reset_index(drop=True)
assert len(t)==240 and t.code.nunique()==40 and not t.duplicated(['code','year']).any()
assert (t.lagyear==t.year-1).all() and (t[['D','lagD','자산총계','lagA','이자비용','매출액']]>0).all().all()
assert t.Sa.between(0,100).all()
t['Post']=t.year.isin([2022,2023]).astype(int)
for v in ['Q', 'S', 'L']:
    t[v + 'P'] = t[v] * t.Post
    t[v + 'T'] = t[v] * (t.year == 2021)
assert np.isfinite(t[['cost','cost_end','burden','margin','icr','Q','S','L']]).all().all()
chk=t.merge(orig[['code','year','L','Y']],on=['code','year'],suffixes=('','_old'),validate='one_to_one')
np.testing.assert_allclose(chk.L, chk.L_old, atol=1e-10)
np.testing.assert_allclose(chk.icr, chk.Y, atol=1e-10)
save(t, 'analysis_panel')
save(av.reset_index(), 'pre_exposures')
save(av.corr().reset_index(), 'pre_correlation')

# 기존 차입의존도와 대조
audit=t[['code','industry','year','D','Qa','차입의존도','cost','cost_end','icr','icr_amount']].copy()
audit['debt_equity'] = 100 * t.D / t['자본총계']
audit['diff_asset'] = audit.Qa - audit['차입의존도']
audit['diff_equity'] = audit.debt_equity - audit['차입의존도']
save(audit, 'definition_audit')
display(audit[['diff_asset', 'diff_equity']].abs().describe())
display(t[['cost','cost_end','burden','margin','icr']].describe())
save(t.nlargest(10,'cost')[['industry','year','D','lagD','이자비용','cost','cost_end']],'largest_cost_proxy')
annual = t.groupby('year')[['cost', 'burden', 'margin', 'icr']].agg(['mean', 'median'])
annual.columns = ['_'.join(c) for c in annual.columns]
save(annual.reset_index(), 'annual_descriptive')
```

### 출력에서 읽을 수 있는 것

기존 차입의존도와 직접 계산한 비율의 절대차 평균은 자산 분모에서 17.723, 자본 분모에서 0.028이었다. 수치상 가까운 분모를 찾은 것이며 공식 정의는 추가 확인이 필요하다.

- 기업 재무 항목을 업종·연도 기준으로 연결했다.
- 네 차입 항목의 합과 그중 단기 몫을 구분했다. 이자비용을 차입 평균잔액으로 나눈 값은 비용 대용률, 전기 자산으로 나눈 값은 자산 대비 부담이다.
- 40개 업종 240행에서 전기 자료가 바로 전년도인지, 단기 몫이 0~100% 안인지 확인했다.
- 기존 차입의존도와 직접 만든 비율을 대조하자 기존 값은 자산보다 자본을 분모로 한 비율에 수치상 가까웠다. 공식 산식은 이 비교만으로 확정되지 않는다.

```python
# 공통 추정식 정의
def design(d,terms):
    X=pd.concat([pd.DataFrame({'const':np.ones(len(d))},index=d.index),d[terms],pd.get_dummies(d.code,prefix='i',drop_first=True,dtype=float),pd.get_dummies(d.year,prefix='y',drop_first=True,dtype=float)],axis=1).astype(float)
    assert np.linalg.matrix_rank(X)==X.shape[1]
    return X
fits = {}
rows = []
def fit(label, d, terms, y='Y'):
    X = design(d, terms)
    model = sm.OLS(d[y].astype(float), X)
    res = model.fit(cov_type='cluster', cov_kwds={'groups': d.code, 'use_correction': True, 'df_correction': True}, use_t=True)
    G = d.code.nunique()
    q = stats.t.ppf(0.975, G - 1)
    for term in terms:
        b = float(res.params[term])
        se = float(res.bse[term])
        rows.append(dict(model=label, outcome=y, term=term, beta=b, se=se, lo=b - q * se, hi=b + q * se, p=float(2 * stats.t.sf(abs(b / se), G - 1)), N=len(d), G=G))
    fits[label] = (d.copy(), terms, model, res)
    return res

# 결과변수별 회귀와 다중검정
terms=['QP','SP','LP','QT','ST','LT']
for y in ['cost', 'burden', 'margin', 'icr']:
    fit('Main_' + y, t, terms, y)
fit('Sensitivity_cost_end',t,terms,'cost_end')
results = pd.DataFrame(rows)
save(results, 'coefficients')
focus=results[results.model.str.startswith('Main_') & results.term.isin(['QP','SP'])].copy()
focus['p_holm8'] = multipletests(focus.p, method='holm')[1]
save(focus, 'focus')
display(focus)

# 업종 제외 민감도 점검
# 독립 OLS/CR1 검산
dd, _, model, res = fits['Main_cost']
X = model.exog
y = model.endog
b = np.linalg.lstsq(X, y, rcond=None)[0]
u = y - X @ b
bread = np.linalg.inv(X.T @ X)
sc = np.array([X[t.code == g].T @ u[t.code == g] for g in sorted(t.code.unique())])
cov = bread @ sc.T @ sc @ bread * 40 / 39 * (len(t) - 1) / (len(t) - X.shape[1])
np.testing.assert_allclose(cov,res.cov_params(),atol=1e-7)
loo=[]
for g in sorted(t.code.unique()):
    dd = t[t.code != g]
    rr = sm.OLS(dd.cost, design(dd, terms)).fit()
    loo.append({'excluded_code': g, 'excluded_industry': t.loc[t.code == g, 'industry'].iloc[0], 'QP': rr.params.QP, 'SP': rr.params.SP})
save(pd.DataFrame(loo),'cost_leave_one_industry_out')
print('full model rank', X.shape[1], '; CR1 verified')
```

### 출력에서 읽을 수 있는 것

Main_margin의 QP는 일반 p=0.015지만 p_holm8=0.122다. 여덟 가지 관계를 살펴본 점까지 고려하면 이 결과 하나를 확정적으로 제시하기 어렵다.

- 같은 사전 재무구조를 쓰면서 결과변수를 바꾸도록 공통 회귀식을 정의했다.
- 총차입 규모와 영업이익률의 개별 관계는 p=0.015였지만, 여덟 비교를 고려하면 p=0.122였다. 이 조합 하나를 강한 증거로 선택하기 어렵다.
- 비용률 모형의 수치를 독립 계산과 대조하고 업종을 하나씩 제외한 결과를 저장했다.

```python
# 부트스트랩과 다중검정
B = 9999
SEED = 20261103
def within_boot(model, groups, param, label):
    xx = pd.DataFrame(model.exog, columns=model.exog_names)
    yy = pd.Series(model.endog)
    gg = pd.Series(np.asarray(groups))
    xx = xx - xx.groupby(gg).transform('mean')
    yy = yy - yy.groupby(gg).transform('mean')
    xx = xx.loc[:, np.max(np.abs(xx), axis=0) > 1e-10]
    assert np.linalg.matrix_rank(xx) == xx.shape[1]
    mm = sm.OLS(yy, xx)
    np.testing.assert_allclose(mm.fit().params[param], model.fit().params[param], atol=1e-08)
    ans = []
    for variant in ['31']:
        res = wildboottest(mm, param=param, cluster=pd.factorize(gg)[0], B=B, seed=SEED, bootstrap_type=variant, impose_null=True, weights_type='rademacher', show=False)
        ans.append({'model': label, 'term': param, 'bootstrap_type': variant, 'p_boot': float(res['p-value'].iloc[0]), 'B': B, 'seed': SEED})
    return ans

boot=[]
for label in ['Main_cost', 'Main_burden', 'Main_margin', 'Main_icr']:
    for term in ['QP', 'SP']:
        dd, _, model, res = fits[label]
        boot.extend(within_boot(model, dd.code, term, label))
boot = pd.DataFrame(boot)
boot['p_boot_holm8'] = multipletests(boot.p_boot, method='holm')[1]
save(boot, 'bootstrap')
display(boot)
```

### 출력에서 읽을 수 있는 것

같은 QP·영업이익률 조합의 부트스트랩 p값은 0.025, 여덟 검정 보정 후에는 0.198이다. 마지막 두 p값 열은 서로 다른 질문에 답한다.

```python
# 기간평균 금액의 변화 분해
a=t[t.year.between(2018,2020)].groupby(['code','industry'])[['영업이익','이자비용','icr','cost','burden','margin']].mean()
b=t[t.year.isin([2022,2023])].groupby(['code','industry'])[['영업이익','이자비용','icr','cost','burden','margin']].mean()
dec = pd.DataFrame(index=a.index)
A0 = a['영업이익']
A1 = b['영업이익']
B0 = a['이자비용']
B1 = b['이자비용']
dec['reconstructed_pre'] = A0 / B0
dec['reconstructed_post'] = A1 / B1
dec['reconstructed_change'] = A1 / B1 - A0 / B0
dec['profit_contribution'] = 0.5 * (A1 - A0) * (1 / B0 + 1 / B1)
dec['interest_contribution'] = 0.5 * (A0 + A1) * (1 / B1 - 1 / B0)
np.testing.assert_allclose(dec.profit_contribution+dec.interest_contribution,dec.reconstructed_change,atol=1e-10)

# 업종별 기여 크기 비교
for v in ['icr', 'cost', 'burden', 'margin']:
    dec[v + '_change'] = b[v] - a[v]
dec['profit_negative_either_period']=(A0<0)|(A1<0)
dec['opposing_contributions']=dec.profit_contribution*dec.interest_contribution<0
dec['profit_contribution_larger_abs']=dec.profit_contribution.abs()>dec.interest_contribution.abs()
dec = dec.reset_index()
save(dec, 'accounting_decomposition')
des = pd.DataFrame([{'industries': len(dec), 'opposing_contributions': int(dec.opposing_contributions.sum()), 'profit_larger_abs': int(dec.profit_contribution_larger_abs.sum()), 'cost_increased': int((dec.cost_change > 0).sum()), 'burden_increased': int((dec.burden_change > 0).sum()), 'icr_declined': int((dec.icr_change < 0).sum()), 'cost_up_and_icr_not_down': int(((dec.cost_change > 0) & (dec.icr_change >= 0)).sum()), 'negative_profit_either_period': int(dec.profit_negative_either_period.sum())}])
save(des, 'descriptive_counts')
display(des)
# 큰 변화를 포함해 전 업종을 표로 공개한다. 대표 사례를 유의성으로 선택하지 않는다.
display(dec[['industry','icr_change','cost_change','profit_contribution','interest_contribution']].sort_values('industry'))
print('Accounting identity verified; all analyses finished')
```

## 결론부터
이자부담 증가 자체는 자료에서 폭넓게 관찰되지만, 그 차이를 사전 단기차입 비중·총차입 규모로 설명하는 회귀 근거는 충분하지 않았다. 40업종 중 38업종의 부담금리 대용지표가 사전기간보다 높아졌고, 공시 ICR이 하락한 업종은 24개였다. 회계적 분해에서는 24업종에서 영업이익 증가가 이자비용 증가의 부정적 기여를 일부 또는 전부 상쇄했다. 공통 부담 증가와 업종 간 차별적 취약성은 서로 다른 질문이다.

## 새로 확인한 세 가지
1. 제공 차입 네 항목의 평균잔액 대비 이자비용은 2018–20 평균 대비 2022–23 평균에 38/40업종에서 상승했다. 업종별 기간평균 지표의 중앙값은 4.14%에서 5.16%로 높아졌다. 같은 두 업종이 중앙값을 구성한다는 뜻은 아니며, 업종별 변화의 중앙값은 +1.09%p다. 이는 기준금리의 인과효과 추정이 아닌 자료의 전후 변화다.

## 바꾼 가설과 회귀 결과
H1: 총차입 규모·유동성을 고려하면 사전 단기항목 비중이 높은 업종에서 부담금리 대용지표가 더 상승하는가? SP 계수 +0.02065, 95% CI [-0.01700, 0.05830], CR1 p=0.2741, 군집 부트스트랩 p=0.4148. 예상한 방향이지만 근거가 충분하지 않다.

RQ3: 수익성·ICR은 어떻게 연결되는가? 총차입/자산과 영업이익률 변화에는 양의 관계가 관찰됐다: QP=+0.5314, 95% CI [0.1077, 0.9552], CR1 p=0.0153, 부트스트랩 p=0.0247. 하지만 4결과×2차입특성의 8개 비교에 Holm 보정을 적용하면 각각 0.1224, 0.1976이다. 따라서 ‘새 가설이 증명됐다’고 강조하지 않고 추가 확인이 필요한 탐색적 신호로 보고한다. 차입이 수익성을 높였다는 인과적 해석도 불가능하다. 단기 비중 및 총차입/자산의 ICR 변화 계수는 각각 +0.1550(p=0.3350), +0.5047(p=0.4289)로 뚜렷하지 않았다.

계수는 사전 재무비율 1%p 차이에 따른 사후 대비의 결과변수 변화 차이다. cost·burden·margin 결과의 단위는 %p, ICR은 배다. 예를 들어 H1의 +0.02065는 단기 비중 10%p 차이에 +0.2065%p의 상대 부담금리 변화 추정에 해당한다. 확정된 효과 크기라고 읽지 않는다. 여러 결과를 본 뒤 골라낸 발견이 되지 않도록 8개 계수와 원 p값·보정값을 모두 제공한다.

## 반드시 남겨야 할 측정 한계
원파일 메타정보와 헤더에서 NICE평가정보/ValueSearch 출처와 십억원 단위는 확인했다. 그러나 항목별 포함 범위·집계 방식은 확인되지 않았다. 사용자는 추가 설명을 보유하지 않는다고 답했다. 따라서 총차입은 ‘제공된 단기차입·단기사채·장기차입·장기사채 합계’, 부담금리는 ‘해당 합계의 기초·기말 평균 대비 이자비용’이라고 한정한다. 공식 실제 약정금리나 정확한 전체 이자부채로 부르지 않는다. 리스부채·유동성장기부채 등의 포함 여부와 항목 중복 여부는 공식 확인이 필요하다.

제공 차입의존도 열은 차입금/자산보다 차입금/자본에 수치상 가깝다. 2018–23년 240행에서 자산 분모와의 평균 절대 차이는 17.72%p, 자본 분모와는 0.028%p였다. 이것은 재계산 결과이지 공식 정의를 확인했다는 뜻이 아니다. 이번에는 해당 비율 열을 사용하지 않고 금액으로 새로 계산했다. 기존 M1–M5의 E도 금액/자산으로 직접 계산했으므로 이 발견이 기존 E 산식을 바꾼 것은 아니다.

금액이 정수 십억원으로 제공되므로 작은 분모에서는 오차가 커질 수 있다. 기말 차입 잔액을 분모로 쓰면 담배 2022년 지표가 47.37%로 커진다. 기초·기말 평균으로 계산한 주 대용지표는 그 해 8.37%다. 이는 평균잔액 사용의 계산상 차이이지, 실제 약정금리의 정답을 확인한 것이 아니다.

## 필요한 범위의 강건성 확인
부담금리 분모를 기말잔액으로 바꾼 모형에서도 단기 비중 계수는 +0.0824(p=0.2382), 총차입/자산 계수는 -0.2273(p=0.2912)로 핵심 방향의 통계적 근거가 뚜렷하지 않았다. 업종 하나씩 제외한 평균잔액 cost 모형에서 SP 계수 범위는 +0.00466~+0.02549, QP는 -0.08752~-0.02106이다. 이 범위는 계수의 민감도이며 모든 제외표본의 유의성을 검정한 것은 아니다. Q와 L의 사전 상관은 -0.749이므로 둘의 조건부 차이를 정밀하게 구분하기 어려울 가능성도 남는다. 이를 이유로 유동비율을 임의로 제거하지 않았다.

## 논문에서 사용할 수 있는 서술과 아직 못 하는 주장
사용 가능한 잠정 서술: ‘제공 차입 항목을 이용한 비용 대용지표는 대부분 업종에서 상승했으나, ICR 변화는 일률적이지 않았다. 영업이익과 이자비용의 회계적 기여가 상쇄되는 업종이 다수였으며, 사전 차입 규모·단기 비중에 따른 차별적 악화의 회귀 근거는 제한적이었다.’
불가능한 주장: ‘고금리가 ICR을 이만큼 낮췄다’, ‘차입이 많으면 실적이 좋아진다’, ‘유동성이 완충한다는 인과효과를 입증했다’, ‘신규 부담금리가 공식 차입금리다’. 또한 통계적 유의성 대신 산업별 차이를 설명하는 결과를 중심으로 재구성했다는 점과 탐색 분석임을 명시해야 한다.

## 다음 단계 및 미확인 자료
현재 자료로 승인된 탐색 계산·회귀·분해는 완료했다. 신규 지표의 제출용 확정 해석은 원출처의 항목 정의 확인이 남아 있다. 제공기관에 확인할 사항은 네 차입 항목의 상호배타성·포함범위, 금액과 비율의 집계 방식·대상 기업, 차입의존도 분모다. 분석을 더 확장하기 전에 이 측정 확인이 우선이다. 전 업종 생산·출하 통제를 하려면 2017–2023년 원지수와 메타정보가 필요하며 아직 반영되지 않았다.
